## Naive Benchmark Comparison

Most organisations budget energy costs by carrying forward the previous year's figures, adjusted at most by a broad inflation assumption. This section compares the forecasting model against that approach, represented here as a seasonal naive model that sets each future value equal to the observed value from the same period one year prior.

The comparison is expressed in dollar terms rather than model error metrics alone, because a percentage improvement in RMSE is not a number a CFO can act on.
